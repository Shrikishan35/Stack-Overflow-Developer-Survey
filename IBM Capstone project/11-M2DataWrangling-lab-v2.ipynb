## Tasks


#### Step 1: Import the necessary module.


### 1. Load the Dataset


<h5>1.1 Import necessary libraries and load the dataset.</h5>


Ensure the dataset is loaded correctly by displaying the first few rows.


```python
# Import necessary libraries
import pandas as pd

# Load the Stack Overflow survey data
dataset_url = "https://cf-courses-data.s3.us.cloud-object-storage.appdomain.cloud/n01PQ9pSmiRX6520flujwQ/survey-data.csv"
df = pd.read_csv(dataset_url)

# Display the first few rows
print(df.head())
```

#### 2. Explore the Dataset


<h5>2.1 Summarize the dataset by displaying the column data types, counts, and missing values.</h5>


```python
# Write your code here
# Data types
print("Data Types:")
print(df.dtypes)

# Non-null counts
print("\nNon-null Counts:")
print(df.count())

# Missing values
print("\nMissing Values:")
print(df.isnull().sum())
```

<h5>2.2 Generate basic statistics for numerical columns.</h5>


```python
# Write your code here
# Basic statistics for numerical columns
print(df.describe())
```

### 3. Identifying and Removing Inconsistencies


<h5>3.1 Identify inconsistent or irrelevant entries in specific columns (e.g., Country).</h5>


```python
# Write your code here
# Check unique Country values
country_values = df["Country"].value_counts(dropna=False)

print(country_values)
```

<h5>3.2 Standardize entries in columns like Country or EdLevel by mapping inconsistent values to a consistent format.</h5>


```python
## Write your code here
# Check for leading/trailing spaces in Country
df["Country"] = df["Country"].str.strip()

# Check the unique Country values again
print(df["Country"].value_counts(dropna=False).head(20))
```

### 4. Encoding Categorical Variables


<h5>4.1 Encode the Employment column using one-hot encoding.</h5>


```python
## Write your code here
# One-hot encode the Employment column
employment_encoded = pd.get_dummies(
    df["Employment"],
    prefix="Employment"
)

# Add the encoded columns to the DataFrame
df = pd.concat([df, employment_encoded], axis=1)

# Display the result
print(df.head())
```

### 5. Handling Missing Values


<h5>5.1 Identify columns with the highest number of missing values.</h5>


```python
## Write your code here
# Count missing values in each column
missing_values = df.isnull().sum()

# Sort from highest to lowest
missing_values = missing_values.sort_values(ascending=False)

# Display top 10
print("Columns with the highest missing values:")
print(missing_values.head(10))
```

<h5>5.2 Impute missing values in numerical columns (e.g., `ConvertedCompYearly`) with the mean or median.</h5>


```python
## Write your code here
# Find the median compensation
median_comp = df["ConvertedCompYearly"].median()

# Fill missing values with the median
df["ConvertedCompYearly"] = df["ConvertedCompYearly"].fillna(median_comp)

# Check again
print(
    "Missing ConvertedCompYearly values:",
    df["ConvertedCompYearly"].isnull().sum()
)
```

<h5>5.3 Impute missing values in categorical columns (e.g., `RemoteWork`) with the most frequent value.</h5>


```python
## Write your code here
# Find the most frequent RemoteWork value
most_frequent_remote = df["RemoteWork"].mode()[0]

print("Most frequent RemoteWork value:",
      most_frequent_remote)

# Fill missing values
df["RemoteWork"] = df["RemoteWork"].fillna(
    most_frequent_remote
)

# Verify
print(
    "Missing RemoteWork values after imputation:",
    df["RemoteWork"].isnull().sum()
)
```

### 6. Feature Scaling and Transformation


<h5>6.1 Apply Min-Max Scaling to normalize the `ConvertedCompYearly` column.</h5>


```python
## Write your code here
# Find minimum and maximum values
min_value = df["ConvertedCompYearly"].min()
max_value = df["ConvertedCompYearly"].max()

# Apply Min-Max scaling
df["ConvertedCompYearly_MinMax"] = (
    (df["ConvertedCompYearly"] - min_value)
    / (max_value - min_value)
)

# Display result
print(
    df[
        ["ConvertedCompYearly",
         "ConvertedCompYearly_MinMax"]
    ].head()
)
```

<h5>6.2 Log-transform the ConvertedCompYearly column to reduce skewness.</h5>


```python
## Write your code here
import numpy as np

# Apply log transformation
df["ConvertedCompYearly_Log"] = np.log1p(
    df["ConvertedCompYearly"]
)

# Display result
print(
    df[
        ["ConvertedCompYearly",
         "ConvertedCompYearly_Log"]
    ].head()
)
```

### 7. Feature Engineering


<h5>7.1 Create a new column `ExperienceLevel` based on the `YearsCodePro` column:</h5>


```python
## Write your code here
# Convert YearsCodePro to numeric
years = pd.to_numeric(df["YearsCodePro"], errors="coerce")

# Create ExperienceLevel
def experience_level(years):
    if pd.isna(years):
        return "Unknown"
    elif years < 3:
        return "Beginner"
    elif years < 7:
        return "Intermediate"
    else:
        return "Experienced"

df["ExperienceLevel"] = years.apply(experience_level)

# Check the result
print(df[["YearsCodePro", "ExperienceLevel"]].head(10))
```
