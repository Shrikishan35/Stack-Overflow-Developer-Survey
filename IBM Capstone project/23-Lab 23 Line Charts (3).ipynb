### Task 1: Trends in Compensation Over Age Groups


##### 1. Line Chart of Median `ConvertedCompYearly` by Age Group


- Track how the median yearly compensation (ConvertedCompYearly) changes across different age groups.

- Use a line chart to visualize these trends.


```python
## Write your code here
# Calculate median compensation by age group
age_comp = df.groupby("Age")["ConvertedCompYearly"].median()

# Create line chart
plt.figure(figsize=(10, 6))

plt.plot(
    age_comp.index,
    age_comp.values,
    marker="o"
)

plt.title("Median Yearly Compensation by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Median Yearly Compensation")
plt.xticks(rotation=45)

plt.grid(True)
plt.show()
```

##### 2. Line Chart of Median `ConvertedCompYearly` for Ages 25 to 45


For a closer look, plot a line chart focusing on the median compensation for respondents between ages 25 and 45.


```python
## Write your code here
# Filter respondents between ages 25 and 45
df_25_45 = df[
    df["Age"].isin([
        "25-34 years old",
        "35-44 years old"
    ])
].copy()

# Calculate median compensation by age group
comp_25_45 = (
    df_25_45
    .groupby("Age")["ConvertedCompYearly"]
    .median()
)

# Create line chart
plt.figure(figsize=(10, 6))

plt.plot(
    comp_25_45.index,
    comp_25_45.values,
    marker="o"
)

plt.title("Median Yearly Compensation for Ages 25 to 45")
plt.xlabel("Age Group")
plt.ylabel("Median Yearly Compensation")
plt.xticks(rotation=45)

plt.grid(True)
plt.show()
```

### Task 2: Trends in Job Satisfaction by Experience Level



##### 1. Line Chart of Job Satisfaction (`JobSatPoints_6`) by Experience Level



- Use a column that approximates experience level to analyze how job satisfaction changes with experience.

- If needed, substitute an available experience-related column for `Experience`.


```python
## Write your code here
# Convert YearsCodePro to numeric values
years_pro = df["YearsCodePro"].replace({
    "Less than 1 year": 0.5,
    "More than 50 years": 51
})

years_pro = pd.to_numeric(years_pro, errors="coerce")

# Create a temporary dataframe
experience_data = pd.DataFrame({
    "YearsCodePro": years_pro,
    "JobSatPoints_6": pd.to_numeric(
        df["JobSatPoints_6"],
        errors="coerce"
    )
}).dropna()

# Calculate median job satisfaction by experience
job_sat_exp = (
    experience_data
    .groupby("YearsCodePro")["JobSatPoints_6"]
    .median()
)

# Create line chart
plt.figure(figsize=(10, 6))

plt.plot(
    job_sat_exp.index,
    job_sat_exp.values,
    marker="o"
)

plt.title("Job Satisfaction by Professional Coding Experience")
plt.xlabel("Years of Professional Coding Experience")
plt.ylabel("Median Job Satisfaction Points")
plt.grid(True)

plt.show()
```

### Task 3: Trends in Job Satisfaction and Compensation by Experience


##### 1.Line Chart of Median ConvertedCompYearly Over Experience Level

- This line chart will track how median compensation (`ConvertedCompYearly`) changes with increasing experience.

- Use a column such as `WorkExp` or another relevant experience-related column.


```python
## Write your code here
# Calculate median compensation by work experience
experience_comp = (
    df.groupby("WorkExp")["ConvertedCompYearly"]
    .median()
    .dropna()
    .sort_index()
)

# Create line chart
plt.figure(figsize=(10, 6))

plt.plot(
    experience_comp.index,
    experience_comp.values,
    marker="o"
)

plt.title("Median Yearly Compensation Over Experience Level")
plt.xlabel("Work Experience")
plt.ylabel("Median Yearly Compensation")

plt.grid(True)
plt.show()
```

##### 2.Line Chart of Job Satisfaction (`JobSatPoints_6`) Across Experience Levels

- Create a line chart to explore trends in job satisfaction (`JobSatPoints_6`) based on experience level.

- This chart will provide insight into how satisfaction correlates with experience over time


```python
## Write your code here
# Calculate median job satisfaction by work experience
experience_sat = (
    df.groupby("WorkExp")["JobSatPoints_6"]
    .median()
    .dropna()
    .sort_index()
)

# Create line chart
plt.figure(figsize=(10, 6))

plt.plot(
    experience_sat.index,
    experience_sat.values,
    marker="o"
)

plt.title("Job Satisfaction Across Experience Levels")
plt.xlabel("Work Experience")
plt.ylabel("Median Job Satisfaction")

plt.grid(True)
plt.show()
```
